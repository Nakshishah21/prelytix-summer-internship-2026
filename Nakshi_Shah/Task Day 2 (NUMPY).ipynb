### BROACASTING

##### Add [1,2,3,4] to every row and subtract [10,20,30,40,50] from every column.


```python
import numpy as np

# Create 5x4 random matrix
matrix = np.random.randint(1, 20, (5, 4))

print("Original Matrix:")
print(matrix)

# Add to every row
row_add = np.array([1, 2, 3, 4])

# Subtract from every column
col_sub = np.array([10, 20, 30, 40, 50]).reshape(5, 1)

# Broadcasting operations
result = matrix + row_add - col_sub

print("\nResult Matrix:")
print(result)
```

#####  Normalise each column of a 6×4 random matrix to have mean=0 and std=1 using broadcasting.

```python
import numpy as np

# Generate 6x4 random matrix
matrix = np.random.randint(1, 100, (6, 4))

print("Original Matrix:")
print(matrix)

# Column-wise mean and std
mean = np.mean(matrix, axis=0)
std = np.std(matrix, axis=0)

# Normalization using broadcasting
normalized = (matrix - mean) / std

print("\nNormalized Matrix:")
print(normalized)

print("\nColumn Means:")
print(np.mean(normalized, axis=0))

print("\nColumn Standard Deviations:")
print(np.std(normalized, axis=0))
```

#####  Task 4.3.3: From two random integer arrays of size 15 (range 1–20), find common elements, elements only in A, and elements only in B. 

```python
import numpy as np

# Create two random arrays
A = np.random.randint(1, 21, 15)
B = np.random.randint(1, 21, 15)

print("Array A:\n", A)
print("\nArray B:\n", B)

# Common elements
common = np.intersect1d(A, B)

# Elements only in A
only_A = np.setdiff1d(A, B)

# Elements only in B
only_B = np.setdiff1d(B, A)

print("\nCommon elements:", common)
print("Elements only in A:", only_A)
print("Elements only in B:", only_B)
```

#####   Task 5.3.2: Simulate a dataset of 50 students with 4 subject marks. Save to CSV. Reload and compute: class topper, subject-wise averages, and students scoring below 50 in any subject.

```python
import numpy as np

# Simulate dataset: 50 students with 4 subject marks
marks = np.random.randint(35, 101, (50, 4))

# Save dataset to CSV
np.savetxt("student_marks.csv", marks, delimiter=",", fmt="%d")

# Reload dataset
data = np.loadtxt("student_marks.csv", delimiter=",")

print("Student Marks Dataset:\n")
print(data)

# Total marks of each student
total_marks = np.sum(data, axis=1)

# Class topper
topper_index = np.argmax(total_marks)

print("\nClass Topper:")
print("Student Index:", topper_index)
print("Marks:", data[topper_index])
print("Total:", total_marks[topper_index])

# Subject-wise averages
subject_avg = np.mean(data, axis=0)

print("\nSubject-wise Averages:")
print(subject_avg)

# Students scoring below 50 in any subject
below_50 = np.where(np.any(data < 50, axis=1))[0]

print("\nStudents scoring below 50 in any subject:")
print(below_50)
```

### Performance & Vectorisation

#####   Task 6.3.1: Rewrite the following loop using vectorisation: result = []; for i in range(len(x)): result.append(x[i]**2 + 2*x[i] + 1). Benchmark both. 

```python
import numpy as np
import time

# Create random array
x = np.random.rand(1000000)

# Using loop
start = time.time()

result_loop = []
for i in range(len(x)):
    result_loop.append(x[i]**2 + 2*x[i] + 1)

end = time.time()
loop_time = end - start

# Using vectorisation
start = time.time()

result_vector = x**2 + 2*x + 1

end = time.time()
vector_time = end - start

print("Loop Time:", loop_time)
print("Vectorised Time:", vector_time)

# Verify both results are same
print("\nResults equal?")
print(np.allclose(result_loop, result_vector))
```

##### Task 6.3.2: Use np.einsum to compute the outer product of two vectors and verify it matches np.outer. 

```python

# Create vectors
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Outer product using einsum
outer_einsum = np.einsum('i,j->ij', a, b)

# Outer product using np.outer
outer_np = np.outer(a, b)

print("Outer Product using einsum:\n", outer_einsum)

print("\nOuter Product using np.outer:\n", outer_np)

# Verify equality
print("\nAre both equal?")
print(np.array_equal(outer_einsum, outer_np))
```

#####   Task 6.3.3: Generate a float32 and float64 array of 1M elements. Compare memory usage and computation speed for squaring each.

```python
import numpy as np
import time

# Create arrays
arr32 = np.random.rand(1000000).astype(np.float32)
arr64 = np.random.rand(1000000).astype(np.float64)

# Memory usage
print("float32 Memory:", arr32.nbytes / (1024**2), "MB")
print("float64 Memory:", arr64.nbytes / (1024**2), "MB")

# Benchmark float32
start = time.time()
sq32 = arr32 ** 2
end = time.time()

time32 = end - start

# Benchmark float64
start = time.time()
sq64 = arr64 ** 2
end = time.time()

time64 = end - start

print("\nfloat32 Squaring Time:", time32)
print("float64 Squaring Time:", time64)
```
