# NLP Preprocessing Tasks

```python
import nltk

nltk.download('punkt_tab')
nltk.download('stopwords')
```

## Convert text to lowercase

```python
text='Natural Language Processing'
print(text.lower())
```

## Remove punctuation

```python
import string
text='Hello, World! Welcome to NLP.'
print(text.translate(str.maketrans('','',string.punctuation)))
```

## Remove numbers

```python
import re
text='I have 2 books and 15 pens.'
print(re.sub(r'\d+','',text))
```

## Remove extra whitespace

```python
text='Python    is   awesome.'
print(' '.join(text.split()))
```

## Word tokenization

```python
from nltk.tokenize import word_tokenize
print(word_tokenize('Python is easy to learn.'))
```

## Sentence tokenization

```python
from nltk.tokenize import sent_tokenize
print(sent_tokenize('Python is easy. NLP is interesting. Practice daily.'))
```

## Remove stopwords

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
sw=set(stopwords.words('english'))
text='This is a simple example of NLP.'
print([w for w in word_tokenize(text) if w.lower() not in sw])
```

## Count words

```python
from nltk.tokenize import word_tokenize
print(len(word_tokenize('Python is easy to learn.')))
```

## Count sentences

```python
from nltk.tokenize import sent_tokenize
print(len(sent_tokenize('Python is easy. NLP is fun. Practice daily.')))
```

## Unique words

```python
from nltk.tokenize import word_tokenize
print(set(word_tokenize('Python is easy. Python is powerful.'.lower())))
```

## Average word length

```python
from nltk.tokenize import word_tokenize
w=[x for x in word_tokenize('Python is easy.') if x.isalpha()]
print(sum(map(len,w))/len(w))
```

## Remove special characters

```python
import re
print(re.sub(r'[^A-Za-z0-9 ]','', 'Python@# is $100% awesome!'))
```

## Check stopword

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
text= "Example of removing stopwords from sentence using NLTK"
stop_word= set(stopwords.words('english'))
tokens= word_tokenize(text)
filtered_tokens= [token for token in tokens if token.lower() not in stop_word]
print(filtered_tokens)
```

## Most frequent word

```python
from nltk.tokenize import word_tokenize
from collections import Counter
w=[x.lower() for x in word_tokenize('Python is easy. Python is powerful. Python is popular.') if x.isalpha()]
print(Counter(w).most_common(1))
```

## Preprocessing pipeline

```python
import string
from nltk.tokenize import word_tokenize
text='Hello, World! Welcome to NLP.'
text=text.lower().translate(str.maketrans('','',string.punctuation))
print(word_tokenize(text))
```
