#####   Task 1: Sort Titanic by fare descending. Who are the top 10 highest-paying passengers? Print their name (if available), fare, and class. 

```python
titanic=sns.load_dataset("titanic")
top_10 = titanic.sort_values(by='fare', ascending=False)[ ['who', 'fare', 'class']].head(10)
print("Top 10 Highest-Paying Passengers:")
print(top_10)
```

#####   Task 2: Sort by pclass ascending, then by age descending within each class. Show the youngest in each class.  

```python
sorted_df = titanic.sort_values(
    by=['pclass', 'age'],
    ascending=[True, False]
)

print("\nSorted Data:")
print(sorted_df[['pclass', 'age', 'sex']].head())

# Youngest in each class
youngest_each_class = titanic.loc[
    titanic.groupby('pclass')['age'].idxmin(),
    ['pclass', 'age', 'sex', 'class']
]

print("\nYoungest Passenger in Each Class:")
print(youngest_each_class)
```

#####   Task 3: Add a rank column to a 10-student marks DataFrame (rank 1=best). Try all 5 rank methods and compare results for tied students. 

```python

df = pd.DataFrame({
    'Name': ['Aman', 'Riya', 'Kunal', 'Sneha', 'Rahul',
             'Priya', 'Vikas', 'Neha', 'Arjun', 'Pooja'],
    'Marks': [95, 88, 95, 76, 88,
              92, 76, 92, 85, 88]
})

print("Original DataFrame:\n")
print(df)

methods = ['average', 'min', 'max', 'first', 'dense']

for method in methods:
    df[f'Rank_{method}'] = df['Marks'].rank(
        method=method,
        ascending=False
    )

print("\nDataFrame with Different Rank Methods:\n")
print(df)
```

##### Task 1: On Titanic: compute survival rate by sex, by pclass, and by (sex × pclass) combination. Which group had the highest survival rate? 

```python
df= sns.load_dataset("titanic")
```

```python
survival_by_sex = df.groupby('sex')['survived'].mean()

# Survival rate by passenger class
survival_by_pclass = df.groupby('pclass')['survived'].mean()

# Survival rate by sex and class combination
survival_by_combo = df.groupby(['sex', 'pclass'])['survived'].mean()

print("Survival Rate by Sex:\n")
print(survival_by_sex)

print("\nSurvival Rate by Pclass:\n")
print(survival_by_pclass)

print("\nSurvival Rate by (Sex × Pclass):\n")
print(survival_by_combo)

# Highest survival group
highest_group = survival_by_combo.idxmax()
highest_rate = survival_by_combo.max()

print("\nHighest Survival Group:")
print(highest_group, "->", highest_rate)
```

#####   Task 2: For each passenger class, find the mean, median, and std of fares. Use a single .agg() call. 

```python
fare_stats = df.groupby('pclass')['fare'].agg(
    ['mean', 'median', 'std']
)

print("\nFare Statistics by Passenger Class:\n")
print(fare_stats)
```

##### Task 3: Add a column class_median_age = median age within each pclass using transform. Then add a flag IsOlderThanClassMedian.

```python
df['class_median_age'] = df.groupby('pclass')['age'].transform('median')

# Add boolean flag
df['IsOlderThanClassMedian'] = (
    df['age'] > df['class_median_age']
)

print("\nData with Class Median Age:\n")
print(df[['pclass', 'age',
          'class_median_age',
          'IsOlderThanClassMedian']].head())
```

#####  Task 4: Filter out passenger classes with fewer than 200 passengers. How many passengers remain? 


```python
filtered_df = df.groupby('pclass').filter(
    lambda x: len(x) >= 200
)

print("\nPassengers Remaining After Filtering:")
print(len(filtered_df))
```

#####   Task 5: Group by embarked port and compute: total passengers, survival rate, average fare. Sort by survival rate descending

```python
embarked_stats = df.groupby('embarked').agg(
    total_passengers=('survived', 'count'),
    survival_rate=('survived', 'mean'),
    average_fare=('fare', 'mean')
).sort_values(
    by='survival_rate',
    ascending=False
)

print("\nEmbarked Port Statistics:\n")
print(embarked_stats)
```

#####   Task 1: Create a Students table (id, name, age) and a Grades table (id, subject, marks). Perform all 4 join types and explain the row count difference. 
  

```python

# Students table
students = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'name': ['Aman', 'Riya', 'Kunal', 'Sneha'],
    'age': [20, 21, 19, 22]
})

# Grades table
grades = pd.DataFrame({
    'id': [1, 2, 2, 5],
    'subject': ['Math', 'Science', 'English', 'Math'],
    'marks': [90, 85, 88, 75]
})

print("Students Table:\n")
print(students)

print("\nGrades Table:\n")
print(grades)

# INNER JOIN
inner_join = pd.merge(students, grades, on='id', how='inner')

# LEFT JOIN
left_join = pd.merge(students, grades, on='id', how='left')

# RIGHT JOIN
right_join = pd.merge(students, grades, on='id', how='right')

# OUTER JOIN
outer_join = pd.merge(students, grades, on='id', how='outer')

print("\nINNER JOIN:\n", inner_join)
print("\nLEFT JOIN:\n", left_join)
print("\nRIGHT JOIN:\n", right_join)
print("\nOUTER JOIN:\n", outer_join)

print("\nRow Counts:")
print("Inner:", len(inner_join))
print("Left :", len(left_join))
print("Right:", len(right_join))
print("Outer:", len(outer_join))
```

##### Task 2: You have two monthly sales DataFrames (same columns). Concatenate them into a single yearly DataFrame. Add a Month column before concatenating. 

  

```python

jan_sales = pd.DataFrame({
    'Product': ['Laptop', 'Phone'],
    'Sales': [50000, 30000]
})

feb_sales = pd.DataFrame({
    'Product': ['Laptop', 'Tablet'],
    'Sales': [45000, 20000]
})

# Add month column
jan_sales['Month'] = 'January'
feb_sales['Month'] = 'February'

# Concatenate
yearly_sales = pd.concat(
    [jan_sales, feb_sales],
    ignore_index=True
)

print("\nYearly Sales Data:\n")
print(yearly_sales)
```

##### Task 3: Merge Titanic with a separate DataFrame mapping pclass → class_label ('First','Second','Third'). Verify all rows have a label. 
  

```python


# Load Titanic dataset
titanic = sns.load_dataset('titanic')

# Mapping DataFrame
class_map = pd.DataFrame({
    'pclass': [1, 2, 3],
    'class_label': ['First', 'Second', 'Third']
})

# Merge
titanic_merged = pd.merge(
    titanic,
    class_map,
    on='pclass',
    how='left'
)

print("\nTitanic with Class Labels:\n")
print(titanic_merged[['pclass', 'class_label']].head())

# Verify all rows have labels
missing_labels = titanic_merged['class_label'].isnull().sum()

print("\nMissing Labels:", missing_labels)
```

##### Task 4: Detect duplicate rows after a merge. Use df.duplicated() to find and df.drop_duplicates() to remove them.

```python
 
# Create duplicate rows intentionally
dup_df = pd.concat(  [inner_join, inner_join.iloc[[0]]], ignore_index=True)

print("\nData with Duplicate Row:\n")
print(dup_df)

# Find duplicates
duplicates = dup_df.duplicated()

print("\nDuplicate Flags:\n")
print(duplicates)

# Show duplicate rows
print("\nDuplicate Rows:\n")
print(dup_df[duplicates])

# Remove duplicates

clean_df = dup_df.drop_duplicates()

print("\nAfter Removing Duplicates:\n")
print(clean_df)
```

#####  Task 1: Build a pivot table: mean and std of age grouped by pclass (rows) and survived (columns). Include totals. 


```python
titanic = sns.load_dataset('titanic')
titanic.head()
pivot_age = pd.pivot_table(
    titanic,
    values='age',
    index='pclass',
    columns='survived',
    aggfunc=['mean', 'std'],
    margins=True
)

print(pivot_age)
```

##### Task 2: Use crosstab to show percentage survival by sex AND embarked port. Which combination had the highest survival rate? 
 

```python
survival_percent = pd.crosstab([titanic['sex'], titanic['embarked']], titanic['survived'],  normalize='index') * 100
print(survival_percent)

highest = survival_percent[1].idxmax()
highest_value = survival_percent[1].max()

print("Highest Survival Combination:", highest)
print("Survival Rate:", highest_value)
```

```python
fare_pivot = pd.pivot_table(
    titanic,
    values='fare',
    index='pclass',
    columns='embarked',
    aggfunc='sum'
)

print(fare_pivot)

print(fare_pivot.idxmax(axis=1))
```
