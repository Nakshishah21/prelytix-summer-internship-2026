## Apply Porter Stemming

```python
from nltk.stem import PorterStemmer
ps=PorterStemmer()
words=['running','studies','playing']
print([ps.stem(w) for w in words])
```

## Apply Snowball Stemming

```python
from nltk.stem import SnowballStemmer
ss=SnowballStemmer('english')
words=['running','studies','playing']
print([ss.stem(w) for w in words])
```

## Compare stemming outputs

```python
from nltk.stem import PorterStemmer,SnowballStemmer
ps=PorterStemmer(); ss=SnowballStemmer('english')
for w in ['running','studies','playing']:
 print(w,ps.stem(w),ss.stem(w))
```

## Lemmatize nouns

```python
from nltk.stem import WordNetLemmatizer
l=WordNetLemmatizer();
print(l.lemmatize('cars',pos='n'))
```

## Lemmatize verbs

```python
from nltk.stem import WordNetLemmatizer
l=WordNetLemmatizer();
print(l.lemmatize('running',pos='v'))
```

## POS tag sentence

```python
from nltk import word_tokenize,pos_tag
print(pos_tag(word_tokenize('The quick brown fox jumps over the lazy dog.')))
```

## Count nouns

```python
from nltk import word_tokenize,pos_tag
p=pos_tag(word_tokenize('The cat eats fish annd dog eats meat.'))
print(p)
print(sum(t.startswith('NN') for _,t in p))
```

## Count verbs

```python
from nltk import word_tokenize,pos_tag
p=pos_tag(word_tokenize('The cat eats fish.'))
print(sum(t.startswith('VB') for _,t in p))
```

## Count adjectives

```python
from nltk import word_tokenize,pos_tag
p=pos_tag(word_tokenize('The quick brown fox jumps.'))
print(sum(t.startswith('JJ') for _,t in p))
```

## Count adverbs

```python
from nltk import word_tokenize,pos_tag
p=pos_tag(word_tokenize('She runs very quickly.'))
print(sum(t.startswith('RB') for _,t in p))
```

## Extract nouns

```python
from nltk import word_tokenize,pos_tag
p=pos_tag(word_tokenize('The cat eats fish.'))
print([w for w,t in p if t.startswith('NN')])
```

## Extract verbs

```python
from nltk import word_tokenize,pos_tag
p=pos_tag(word_tokenize('The cat eats fish.'))
print([w for w,t in p if t.startswith('VB')])
```

## POS tag frequencies

```python
from nltk import word_tokenize,pos_tag
from collections import Counter
p=pos_tag(word_tokenize('The quick brown fox jumps over the lazy dog.'))
print(Counter(tag for _,tag in p))
```
