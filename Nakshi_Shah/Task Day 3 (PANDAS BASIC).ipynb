#####  Task 3: From the Series [45,82,67,55,91,38,74], filter all values above the mean. Print both the mask and the filtered values.

#####  Task 1: On the Titanic dataset: (a) find % missing per column, (b) drop columns with >50% missing, (c) fill Age with median grouped by Pclass.  

```python
titanic = sns.load_dataset("titanic")
print("\nNull Values Per Column:")
missing=titanic.isnull().sum()/len(titanic)*100
print(missing)
titanic_clean=titanic.loc[:,missing<=50]
print(titanic_clean)
titanic_clean["age"]=titanic_clean["age"].fillna(titanic_clean.groupby("pclass")["age"].transform("median"))
```

##### Task 3: Use df.fillna(df.mean()) to fill all numeric nulls at once. Then use df.fillna(df.mode().iloc[0]) for non-numeric nulls. 

```python
import numpy as np
data2 = {
    "Age": [22, np.nan, 24, np.nan],
    "Salary": [50000, 60000, np.nan, 55000],
    "City": ["Ahmedabad", np.nan, "Delhi", "Delhi"],
    "Department": ["IT", "HR", np.nan, "IT"]
}

df3 = pd.DataFrame(data2)

print("\nOriginal DataFrame:\n")
print(df3)



numeric_cols = df3.select_dtypes(include=np.number).columns

df3[numeric_cols] = df3[numeric_cols].fillna(
    df3[numeric_cols].mean()
)



non_numeric_cols = df3.select_dtypes(exclude=np.number).columns

df3[non_numeric_cols] = df3[non_numeric_cols].fillna(
    df3[non_numeric_cols].mode().iloc[0]
)

print("\nDataFrame after filling null values:\n")
print(df3)
```

##### Task 1: On Titanic: add FamilySize, IsAlone (1/0), and AgeGroup ('Child'/<18, 'Adult'/18-60, 'Senior'>60). 

```python
import pandas as pd
import numpy as np
import seaborn as sns

# Load Titanic dataset
df = sns.load_dataset("titanic")

# ============================================
# TASK 1
# Add FamilySize, IsAlone, AgeGroup
# ============================================

# FamilySize = siblings/spouses + parents/children + self
df["FamilySize"] = df["sibsp"] + df["parch"] + 1

# IsAlone = 1 if FamilySize == 1 else 0
df["IsAlone"] = np.where(df["FamilySize"] == 1, 1, 0)

# AgeGroup column
df["AgeGroup"] = pd.cut(
    df["age"],
    bins=[0, 18, 60, 100],
    labels=["Child", "Adult", "Senior"]
)

print("Task 1 Output:\n")
print(df[["age", "FamilySize", "IsAlone", "AgeGroup"]].head())
```
