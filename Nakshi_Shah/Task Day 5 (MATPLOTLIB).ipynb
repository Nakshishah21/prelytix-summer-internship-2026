### MatplotLib


```python
import numpy as np
import matplotlib.pyplot as plt
```

##### Task 1: Create a line plot showing the squares of numbers from 1 to 10. Make the line dashed and green. 


```python
x= np.arange(1,15)
y=x**2
plt.plot(x,y,linestyle="--",color="g")
plt.title("Squares of Numbers")
plt.xlim(0,12)
plt.xlabel("Numbers")
plt.ylabel("Squares")
plt.grid(True)
plt.show()
```

##### 

##### Task 2: Plot two lines on the same graph: y=x^2 and y=x^3. Add a legend to distinguish them. 

```python
x= np.arange(1,10)
y1=x**2
y2=x**3
plt.plot(x,y1,linestyle="--",color="g", label="Squares")
plt.plot(x,y2,linestyle="-" , color="r", label="Cubes")

plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.legend(loc=")
plt.show()
```

#####   Task 3: Create a scatter plot of height vs weight using dummy data. Color the points purple and use triangles as markers. 

```python
height = [150, 155, 160, 165, 170, 175, 180]
weight = [45, 50, 55, 60, 68, 72, 80]

plt.figure(figsize=(6,4))

plt.scatter(height, weight, color='purple', marker='*')

plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.grid(True)

plt.show()
```

#####   Task 1: Create a horizontal bar chart (plt.barh) showing population of 5 different countries. 
  

```python
countries = ['India', 'USA', 'China', 'Brazil', 'Canada']
population = [1400, 331, 1441, 213, 38]   # in millions

plt.figure(figsize=(7,4))

plt.barh(countries, population,color=["blue","red","green","m","y"])

plt.title("Population of Countries")
plt.xlabel("Population (millions)")
plt.ylabel("Countries")

plt.show()
```

##### Task 2: Generate 500 random ages between 0 and 80. Plot a histogram with 10 bins. Add a black edge color to the bins. 


```python
ages= np.random.randint(0,81,500)
plt.hist(ages,bins=30,color="b",edgecolor="black")
plt.plot()
plt.title("agesss")
```

#####   Task 3: Create a pie chart showing your daily time allocation (sleep, work, study, leisure). Explode the 'study' slice. 

```python
routine=["sleep","work","study","leisure"]
time=[8,5,5,6]
plt.pie(time, labels=routine,explode=[0,0,0.1,0],  autopct='%1.1f%%', colors=["g","r","y","m"])
plt.plot()
plt.title("My Routine")
```

##### Task 3.2: Create a 2x2 subplot grid. Plot a line, scatter, bar, and histogram in the 4 axes respectively. Add titles to each. 

```python
fig, axes = plt.subplots(2, 2, figsize=(8,6))

# 1. Line Plot
axes[0,0].plot(x, y)
axes[0,0].set_title("Line Plot")

# 2. Scatter Plot
height = [150, 155, 160, 165, 170]
weight = [45, 50, 55, 65, 70]

axes[0,1].scatter(height, weight, color='purple', marker='^')
axes[0,1].set_title("Scatter Plot")

# 3. Bar Plot
countries = ['India', 'USA', 'China', 'Canada']
population = [1400, 331, 1441, 38]

axes[1,0].bar(countries, population)
axes[1,0].set_title("Bar Chart")

# 4. Histogram
ages = np.random.randint(0, 80, 500)

axes[1,1].hist(ages, bins=10, edgecolor='black')
axes[1,1].set_title("Histogram")

plt.tight_layout()

plt.show()
```
