# Sentiment Analysis and Named Entity Recognition Tasks

```python
# Install if needed:
!pip install textblob spacy
!python -m textblob.download_corpora
!python -m spacy download en_core_web_sm

from textblob import TextBlob
import spacy
from collections import Counter
nlp = spacy.load("en_core_web_sm")
```

## Perform sentiment analysis on one sentence

```python
text='I love learning NLP.'
blob=TextBlob(text)
print(blob.sentiment)
```

## Compare sentiment of two sentences

```python
s1=TextBlob('I love this movie.')
s2=TextBlob('I hate this movie.')
print(s1.sentiment,s2.sentiment)
```

## Detect positive words

```python
text='I love this beautiful and amazing place.'
print([w for w in text.split() if TextBlob(w).sentiment.polarity>0])
```

## Detect negative words

```python
text='This is a bad and horrible experience.'
print([w for w in text.split() if TextBlob(w).sentiment.polarity<0])
```

## Extract named entities

```python
doc=nlp('Sundar Pichai is the CEO of Google in California.')
print([(e.text,e.label_) for e in doc.ents])
```

## Identify person names

```python
doc=nlp('Elon Musk founded SpaceX.')
print([e.text for e in doc.ents if e.label_=='PERSON'])
```

## Identify organization names

```python
doc=nlp('Microsoft acquired GitHub.')
print([e.text for e in doc.ents if e.label_=='ORG'])
```

## Identify locations

```python
doc=nlp('I live in Ahmedabad, India.')
print([e.text for e in doc.ents if e.label_ in ['GPE','LOC']])
```

## Extract dates

```python
doc=nlp('The meeting is on 15 August 2026.')
print([e.text for e in doc.ents if e.label_=='DATE'])
```

## Count named entities

```python
doc=nlp('Sundar Pichai works at Google in California.')
print(doc.ents)
print(len(doc.ents))
```

## Check sentence polarity

```python
print(TextBlob('Python is amazing!').sentiment.polarity)
```

## Check subjectivity score

```python
print(TextBlob('Python is amazing!').sentiment.subjectivity)
```

## Find repeated words

```python
text='python python java c java python'
from collections import Counter
c=Counter(text.split())
print([w for w,v in c.items() if v>1])
```

## Detect duplicate sentences

```python
s=['Python is easy.','Python is fun.','Python is easy.']
print(len(s)!=len(set(s)))
```

## Remove duplicate words

```python
text='python python java java c'
print(' '.join(dict.fromkeys(text.split())))
```
