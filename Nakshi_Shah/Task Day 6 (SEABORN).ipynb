### Seaborn 

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

##### Task 1: Load the 'penguins' dataset from Seaborn. Create a scatter plot of bill_length_mm vs bill_depth_mm, colored (hue) by species. 
  

```python
penguins = sns.load_dataset("penguins")
plt.figure(figsize=(7,5))

sns.scatterplot(data=penguins,
                x='bill_length_mm',
                y='bill_depth_mm',
                hue='species')

plt.title("Bill Length vs Bill Depth")
plt.xlabel("Bill Length (mm)")
plt.ylabel("Bill Depth (mm)")

plt.show()
```

##### Task 2: Load the 'flights' dataset. Create a line plot showing the number of passengers over the years. Seaborn will automatically average the months.

```python
flights= sns.load_dataset("flights")
plt.figure(figsize=(8,5))

sns.lineplot(data=flights,
             x='year',
             y='passengers')

plt.title("Passengers Over the Years")
plt.xlabel("Year")
plt.ylabel("Passengers")

plt.grid(True)

plt.show()
```

##### Task 1: Using the 'titanic' dataset, create a histogram of passenger ages. Add a KDE line and hue by 'survived'. 
 

```python
titanic = sns.load_dataset('titanic')
plt.figure(figsize=(8,5))

sns.histplot(data=titanic,
             x='age',
             hue='survived',
             kde=True,
             bins=20 ,palette=
            "dark")

plt.title("Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

##### ` Task 2: Create a countplot showing the number of passengers in each 'pclass' on the Titanic. 
  

```python
titanic = sns.load_dataset('titanic')
plt.figure(figsize=(6,4))

sns.countplot(data=titanic,
              x='pclass', palette="pastel")

plt.title("Number of Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")

plt.show()
```

##### Task 3: Create a boxplot showing the 'fare' distribution across different 'pclass' categories. 

```python
titanic= sns.load_dataset("titanic")
plt.figure(figsize=(7,5))

sns.boxplot(data=titanic,
            x='pclass',
            y='fare',color="r")

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

plt.show()
```

##### Task 1: Calculate the correlation matrix for the 'tips' dataset. Plot it as a heatmap with annotations (annot=True) and a 'viridis' colormap. 


```python
tips = sns.load_dataset('tips')


corr_matrix = tips.corr(numeric_only=True)

plt.figure(figsize=(8,6))

sns.heatmap(corr_matrix,
            annot=True,
            cmap='viridis')

plt.title("Correlation Heatmap - Tips Dataset")

plt.show()
```
